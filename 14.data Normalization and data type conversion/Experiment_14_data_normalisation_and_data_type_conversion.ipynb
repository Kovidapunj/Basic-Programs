```python
#dataset
import pandas as pd

data = {
    'Product':['Laptop','Mobile','Tablet','Headphones','Smartwatch','Camera'],
    'Price':[55000,20000,30000,2000,8000,45000],
    'Unit_Sold':[25,60,35,80,50,15],
    'Discount':[10,15,12,20,18,8],
}
df = pd.DataFrame(data)
df
```

```python
#1. Min-Max Normalisation
df['price_MinMax'] = (df['Price'] - df['Price'].min()) / (df['Price'].max() - df['Price'].min())
print("\nMin-Max Normalised Price")
print(df[['Product','Price','price_MinMax']])
```

```python
#Normalize multiple columns
cols = ['Price','Unit_Sold','Discount']
df_norm = (df[cols] - df[cols].min()) / (df[cols].max() - df[cols].min())
print("\nNormalized dataset")
print(df_norm)
```

```python
#2. Z-Score Normalization
df['Units_Zscore'] = (df['Unit_Sold'] - df['Unit_Sold'].mean()) / df['Unit_Sold'].std()
print("\nZ-Score Normalised Unit_Sold")
print(df[['Product','Unit_Sold','Units_Zscore']])
```

```python
#3. Decimal Scaling
df['Price_Decimal'] = df['Price'] / 100000
print("\nDecimal Scaling")
print(df[['Product','Price','Price_Decimal']])
```

```python
import numpy as np
import pandas as pd

data = {
    'order_id':[101,102,103,104,105,106],
    'Customer_Gender': ['Male','Female','Male','Female','Male','Female'],
    'Product_Category':['Electronics','Clothing','home','Electronics','beauty','Books'],
    'Payment_Method':['UPI','Credit Card','Debit Card','COD','UPI','Debit Card'],
    'City':['Pune','Mumbai','Delhi','Bangalore','Mumbai','Hyderabad'],
    'Order Value':[4500.32,2300,5400,1200,8000,6500],
}
df = pd.DataFrame(data)
df
```

```python
#Method 1 : Label Encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Gender_Label'] = le.fit_transform(df['Customer_Gender'])
print(df)
#works for only 2 different values
```

```python
#Method 2: One-Hot Encoding
df_encoding = pd.get_dummies(df,columns=['Product_Category','Payment_Method','City'])
print(df_encoding)
```

```python
#Method 3: Dummy Encoding (Drop First)
df_dummy = pd.get_dummies(df,columns=['Product_Category','Payment_Method','City'],drop_first=True)
print(df_dummy)
```
